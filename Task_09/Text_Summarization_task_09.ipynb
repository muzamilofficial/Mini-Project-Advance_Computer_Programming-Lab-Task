### Text Summarization

```python
from transformers import T5Tokenizer, T5ForConditionalGeneration
```

```python
checkpoint = "t5-small"
tokenizer = T5Tokenizer.from_pretrained(checkpoint)
seq2seq_model = T5ForConditionalGeneration.from_pretrained(checkpoint)
```

```python
article = '''Knowledge grows when curiosity meets perseverance and an open mind.
Each obstacle encountered on the path to learning is a stepping stone toward greater understanding.
Those who embrace continuous improvement and maintain a clear sense of purpose tend to accomplish far more than those who abandon their efforts too soon.
Building confidence in one's own potential is a key part of achieving lasting success.
Through discipline and genuine passion, any individual can transform their aspirations into meaningful accomplishments.
'''
prompt_text = "summarize: " + article
```

```python
encoded_input = tokenizer(
    prompt_text,
    return_tensors="pt",
    max_length=512,
    truncation=True
)

output_ids = seq2seq_model.generate(
    encoded_input["input_ids"],
    attention_mask=encoded_input["attention_mask"],
    max_length=60,
    min_length=20,
    length_penalty=2.0,
    num_beams=4,
    early_stopping=True
)
```

```python
result_text = tokenizer.decode(output_ids[0], skip_special_tokens=True)

print("Summary:", result_text)
```
